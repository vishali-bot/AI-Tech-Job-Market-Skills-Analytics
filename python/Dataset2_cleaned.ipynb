```python
annual_multiplier = {
    "YEARLY": 1,
    "MONTHLY": 12,
    "BIWEEKLY": 26,
    "WEEKLY": 52,
    "HOURLY": 2080
}

df["Salary_Min_Annual"] = (
    df["min_salary"] *
    df["pay_period"].map(annual_multiplier)
)

df["Salary_Max_Annual"] = (
    df["max_salary"] *
    df["pay_period"].map(annual_multiplier)
)
```

```python
df.loc[df["currency"] != "USD", ["Salary_Min_Annual", "Salary_Max_Annual"]] = pd.NA
```
