```python
#removing duplicate values

def remove_duplicate_locations(value):
    locations = [x.strip() for x in value.split(",")]
    
    unique_locations = []
    for location in locations:
        if location not in unique_locations:
            unique_locations.append(location)
    
    return ", ".join(unique_locations)

df["Locations_Cleaned"] = df["Locations_Cleaned"].apply(
    remove_duplicate_locations
)
```

```python
import re
import pandas as pd

def convert_to_days(value):
    value = str(value).strip().lower()

    # Current postings
    if value in ["just now", "few hours ago", "today"]:
        return 0

    # More than 30 days
    if "30+ days" in value:
        return 31

    # Future postings
    if "starts" in value:
        return pd.NA

    # Unknown
    if value == "unknown":
        return pd.NA

    # Normal "X Days Ago"
    match = re.search(r"(\d+)", value)

    if match:
        return int(match.group(1))

    return pd.NA


df["Days_Ago"] = (
    df["Post_Time"]
    .apply(convert_to_days)
    .astype("Int64")
)
```
